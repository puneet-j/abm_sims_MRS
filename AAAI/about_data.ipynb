```python
import os 
import sys
import pandas as pd 
import numpy as np

folder = './data/multirunsims/'
files = os.listdir(folder)
files = [file for file in files if file.endswith('.csv') and file.startswith('1')]
meta = pd.read_csv(folder+'metadata.csv')
```

```python
submeta = pd.DataFrame([],columns=meta.columns)
submeta
```

```python
import pdb
# files.remove('metadata.csv')
for file in files:
    
    row = meta.loc[meta.file_name==file].values
    # print(row, len(row[0]))
    try:
        submeta.loc[-1] = row[0]  # adding a row
    except:
        pdb.set_trace()
    # break
    submeta.index = submeta.index + 1  # shifting index
    submeta = submeta.sort_index()  # sorting by index
    # break
submeta

```

```python
from ast import literal_eval
submeta.site_qualities=submeta.site_qualities.apply(literal_eval)
submeta.site_positions=submeta.site_positions.apply(literal_eval)
submeta.site_qualities=submeta.site_qualities.apply(lambda x: tuple(x))
submeta.site_positions=submeta.site_positions.apply(lambda x: tuple([tuple(a) for a in x]))

df = submeta.groupby(by=['site_qualities', 'site_positions', 'num_agents'], as_index=False).agg(lambda x: x.tolist())
```

```python
df['numentries'] = df.maxTime.apply(lambda x: len(x))
df.numentries

```
