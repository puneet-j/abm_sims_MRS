```python
import os 
import sys
import pandas as pd 
import numpy as np
folder = './data/potential_fails/'
files = os.listdir(folder)
files = [file for file in files if file.endswith('.csv')]


meta = pd.read_csv('./data/metadata.csv')

counter = 0

```

```python
# file = files[0]
for file in files:
    row = meta[meta.file_name==file].site_converged.to_numpy()
    if np.isnan(row[0]):
        continue
    else:
        counter += 1

print(counter)
```
