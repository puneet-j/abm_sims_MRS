```python
import os 
import sys
import pandas as pd 
import numpy as np

folder = './data/quorumsims_fixed_otheragents/train/'
files = os.listdir(folder)
files = [file for file in files if file.endswith('.csv') and file.startswith('1')]


meta = pd.read_csv(folder + 'metadata.csv')

counter = 0

```

```python
# file = files[0]
for file in files:
    row = meta[meta.file_name==file].site_converged.to_numpy()
    print(row, file)
    if np.isnan(row[0]):
        continue
    else:
        counter += 1

print(counter)
```
